```python

from transformers import pipeline

# Sentiment Analysis
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

text = "I absolutely love this course!"

result = classifier(text)

print(result)
```

```python

from transformers import pipeline

# Text Summarization
summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)

text = """
Artificial Intelligence is transforming the modern world.
It is being used in healthcare, education, finance,
transportation and many other industries.
AI systems can analyze large amounts of data,
automate repetitive tasks and help people make
better decisions. Machine learning and deep learning
are important technologies behind modern AI systems.
"""

summary = summarizer(
    text,
    max_length=60,
    min_length=20,
    do_sample=False
)

print(summary)
```

```python
from transformers import pipeline

# Text Generation
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

prompt = "Artificial Intelligence is changing the world because"

result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1
)

print(result[0]["generated_text"])
```

```python
from transformers import pipeline

# Language Translation: English to French
translator = pipeline(
    "translation_en_to_fr",
    model="Helsinki-NLP/opus-mt-en-fr"
)

text = "Artificial Intelligence is transforming the world."

result = translator(text)

print(result[0]["translation_text"])
```

```python
from transformers import pipeline

# Zero-Shot Classification
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

text = "I love learning Python and building AI projects."

labels = [
    "Technology",
    "Sports",
    "Politics",
    "Education"
]

result = classifier(
    text,
    candidate_labels=labels
)

print(result)
```
